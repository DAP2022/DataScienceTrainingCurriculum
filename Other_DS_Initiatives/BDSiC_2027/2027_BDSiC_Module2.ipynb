### 1a. Assigning Variables

A variable in Python is **a name that indicates where in memory we 'parked' a value**.  

To assign a variable in Python, you need:
* a name (e.g. 'a')
* a value (e.g. '1')

Variable names have restrictions in Python.
* They can contain letters, underscores, and numbers
* They are case sensitive
* They cannot begin with a number

### 1b. Variables as Pointers

When you assign a variable, you are creating a reference to an object.
- *Everything* in Python is an **object**, which is Python's abstraction for data.
- All objects have:
    - **Identity**: a unique identifier that cannot be changed, representing where the object is stored in memory
    - **Type**: the kind of object, which determines what you can do with it
    - **Value**: the data stored in the object, which can be **mutable** (changeable) or **immutable** (unchangeable)
- **Variables serve as pointers to objects**. Whenever you assign a value to a variable, you are pointing to something in memory.
<br>

We already saw examples of value in section 1a. Next, let's look more closely at identity and at variables as pointers.

Follow along with the video for the following code cells to hear an explanation about identity and to get an introduction to [Python Tutor](https://pythontutor.com/visualize.html#mode=edit).

```python
# -------------------------------------------
# Reminder! Hashed out lines are comments and are not interpreted by Python.
# So you can use them to write instructions or relevant information
# to your future self or anyone else who uses your script.
# --------------------------------------------

# Examples of variables and how/where they are stored.
# Copy and paste this cell into Python Tutor to see line
# by line what is happening with memory
p = 100
print("Here is the initial value of p:", p)
q = 100
print("Here is the value of q:", q)
n = p
print("Here is the value of n:", n)
```

```python
# We can see where the value is stored using id()
print('---location of the three variables-----')
print(id(p))
print(id(n))
print(id(q))
```

```python
# TAKE HOME: changing the value associated with p
# changes where p 'points' in memory
p = 50
print("Here is the new value of p:", p, "and it is located here:", id(p))
print("We have changed the value of p, so now what is the value of n:", n,
      "and it is located here:", id(n))
# we can see where the values are stored,
# and the two variables no longer point to the same memory slot
```
