### 1b. Variables as Pointers

When you assign a variable, you are creating a reference to an object.
- *Everything* in Python is an **object**, which is Python's abstraction for data.
- All objects have:
    - **Identity**: a unique identifier that cannot be changed, representing where the object is stored in memory
    - **Type**: the kind of object, which determines what you can do with it
    - **Value**: the data stored in the object, which can be **mutable** (changeable) or **immutable** (unchangeable)
- **Variables serve as pointers to objects**. Whenever you assign a value to a variable, you are pointing to something in memory.
<br>

We already saw examples of value in section 1a. Next, let's look more closely at identity and at variables as pointers.

Follow along with the video for the following code cells to hear an explanation about identity and to get an introduction to [Python Tutor](https://pythontutor.com/visualize.html#mode=edit).

````python
# -------------------------------------------
# Reminder! Hashed out lines are comments and are not interpreted by Python.
# So you can use them to write instructions or relevant information
# to your future self or anyone else who uses your script.
# --------------------------------------------

# Examples of variables and how/where they are stored.
# Copy and paste this cell into Python Tutor to see line
# by line what is happening with memory
p = 100
print("Here is the initial value of p:", p)
q = 100
print("Here is the value of q:", q)
n = p
print("Here is the value of n:", n)
````

````python
# We can see where the value is stored using id()
print('---location of the three variables-----')
print(id(p))
print(id(n))
print(id(q))
````

````python
# TAKE HOME: changing the value associated with p
# changes where p 'points' in memory
p = 50
print("Here is the new value of p:", p, "and it is located here:", id(p))
print("We have changed the value of p, so now what is the value of n:", n,
      "and it is located here:", id(n))
# we can see where the values are stored;
# the two variables no longer point to the same memory slot
````

### 1c. Data Types

Python has built-in data types. Examples include:
* __Numeric__: integer, float, complex
* __Text:__ string
* __Sequence:__ list, tuple, range
* __Mapping:__ dictionary
* __Nonetype__
* __Boolean__
<br>

While we will see most of these during this course, we will first focus on integers, floats, booleans, and strings.

In our examples so far, we've seen both integers (`int`) and strings (`str`). We can use the `type()` function to examine the type of data a variable is pointing to:

````python
hi_1 = "Hello"
print("hi_1 says", hi_1)
print("hi_1 is", type(hi_1))

print("----")

p = 50
print("the value of p is", p)
print("p is", type(p))
````

#### Numeric Data
Within the numeric data types, the types we will see frequently are integer and float.

* Integers are only whole numbers (positive or negative)
* Floats include numbers with a decimal

Both are included because they are needed in different contexts. For example, if you have a list of items, you can identify the item at position 3 in the list, but not at position 3.57. In that context, limiting to whole numbers makes sense.  
<br>

#### Booleans
Boolean (`bool`) is a built-in data type that represents `True` or `False`. Booleans are useful for evaluating whether or not statements are true. We will see much more of Booleans in Module 3.

### 1d. Manipulating Variables
#### Arithmetic Operators
You can manipulate variables using simple math in Python, including:

* addition: +
* subtraction: -
* multiplication: \*
* division: /
* modulus (remainder): %
* floor division (round down after dividing): //
* exponents: \*\*

The following code cell shows an example of basic arithmetic operations on variables.

````python
# You can do simple math in Python:
WaterOxygens = 1
WaterHydrogens = 2
OxygenMass = 15.9994 # standard atomic weight
HydrogenMass = 1.00794

#calculating the mass of a water molecule
#H20 = 2*MassOfHydrogen + 1*MassOfOxygen
WaterMass = WaterOxygens*OxygenMass+WaterHydrogens*HydrogenMass
print("The Molecular weight of a molecule of Water is:",WaterMass)
````
